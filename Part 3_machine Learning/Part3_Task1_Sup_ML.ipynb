```python
df_model= pd.read_csv(r'C:/Users/Sadanand/OneDrive/NUS AI and ML Course\\00 Capstone Project/capstone_projct_shrikant/data/ml_Metro_Interstate_Traffic_Volume.csv')
df_model.head()
```

```python
y = df_model['traffic_volume']
x = df_model[['temp','rain_1h','snow_1h','clouds_all','high_risk']]
```

```python
df_model.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3)
```

```python
lm = LinearRegression()
```

```python
model = lm.fit(x_train,y_train)
pred_train = model.predict(x_train)
pred_test = model.predict(x_test)
```

```python
logger.info('Training MAE: %s', metrics.mean_absolute_error(y_train,pred_train))
logger.info('Testing MAE: %s', metrics.mean_absolute_error(y_test,pred_test))
```

```python
logger.info('Training Rsquared: %s', metrics.r2_score(y_train,pred_train))
logger.info('Testing Rsquared: %s', metrics.r2_score(y_test,pred_test))
```

```python
xreg_train, xreg_test, yreg_train, yreg_test = train_test_split(x,y,test_size=0.3)
```

```python
dt_reg = DecisionTreeRegressor()
model = dt_reg.fit(xreg_train,yreg_train)
pred_train_reg = model.predict(xreg_train)
pred_test_reg = model.predict(xreg_test)
logger.info('Training MAE: %s', metrics.mean_absolute_error(yreg_train,pred_train_reg))
logger.info('Testing MAE: %s', metrics.mean_absolute_error(yreg_test,pred_test_reg))
logger.info('Training Rsquared: %s', metrics.r2_score(yreg_train,pred_train_reg))
logger.info('Testing Rsquared: %s', metrics.r2_score(yreg_test,pred_test_reg))
```

```python
lm_ols = sm.OLS(yreg_train,xreg_train).fit()
lm_ols.summary()
```

```python
comparison_df = pd.DataFrame({
    'Actual Traffic': yreg_test.values,
    'Predicted Traffic': pred_test_reg.round(0)
})

logger.info("Comparison Head (Regression):\n%s", comparison_df.head(10))
```

```python
y_dt = df_model['high_risk']
x_dt = df_model[['temp','rain_1h','snow_1h','clouds_all','hour','is_weekend']]
dt_model = DecisionTreeClassifier()
xdt_train, xdt_test, ydt_train, ydt_test = train_test_split(x_dt,y_dt,test_size=0.3)

dt_model.fit(xdt_train,ydt_train)
predictdt_y = dt_model.predict(xdt_test)
accuracy_dt = dt_model.score(xdt_test,ydt_test)
logger.info("Decision Tree accuracy is : %s", accuracy_dt)
logger.info("Classification Report:\n%s", classification_report(ydt_test, predictdt_y))
```

```python
logger.info("Predictions: %s", predictdt_y[:25])
```
